```python
from src.rag import ask

questions = [
    'What is the difference between a list and a tuple in Python?',
    'How can overfitting be reduced in machine learning?',
    'What is the purpose of a LEFT JOIN?',
    'What is the shared responsibility model in cloud computing?',
    'Why does a RAG system use chunk overlap?',
    'What is the capital of France?'
]

for question in questions:
    print('\nQUESTION:', question)
    answer, results = ask(question)
    print(answer)
    print('Retrieved sources:')
    for item in results:
        location = f"page {item['page']}" if item['page'] else f"chunk {item['chunk_id']}"
        print('-', item['source'], '|', location, '| score=', round(item['score'], 3))
```
